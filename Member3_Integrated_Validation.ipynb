```python

from sklearn.ensemble import HistGradientBoostingRegressor
from threadpoolctl import threadpool_limits
import numpy as np
import pandas as pd

# December 2025
month = pd.Period("2025-12", freq="M")

# Train only on transactions before December
train_mask = listings["date_sold"] < month.start_time

# Validate on transactions sold in December
valid_mask = (
    listings["date_sold"].dt.to_period("M") == month
)

# Hide unavailable information
safe = data.hide_answers(listings, "off_market")

# Use Member 2's new features
X = features.off_market_features(safe)

# Default Gradient Boosting model
model = HistGradientBoostingRegressor(random_state=0)

with threadpool_limits(limits=4):
    model.fit(
        X.loc[train_mask],
        np.log(listings.loc[train_mask, "sold_price"])
    )

    predictions = np.round(
        np.exp(model.predict(X.loc[valid_mask]))
    )

actual = listings.loc[
    valid_mask, "sold_price"
].to_numpy()

ape = np.abs(predictions - actual) / actual * 100

print("Validation month:", month)
print("Number of homes:", len(actual))
print("MdAPE:", round(np.median(ape), 3), "%")
print("MAPE:", round(np.mean(ape), 3), "%")
```

```python
validation_months = pd.period_range(
    "2025-12",
    "2026-05",
    freq="M"
)

print(validation_months)
```

```python
model_configs = {
    "off_market": {
        "default": {
            "random_state": 0
        },
        "tuned_round8": {
            "learning_rate": 0.1,
            "max_iter": 600,
            "max_leaf_nodes": 80,
            "min_samples_leaf": 20,
            "l2_regularization": 0,
            "early_stopping": False,
            "random_state": 0
        }
    },
    "list_aware": {
        "default": {
            "random_state": 0
        },
        "tuned_round8": {
            "learning_rate": 0.1,
            "max_iter": 500,
            "max_leaf_nodes": 45,
            "min_samples_leaf": 20,
            "l2_regularization": 0,
            "early_stopping": True,
            "random_state": 0
        }
    }
}

print("Model configurations ready!")
```

```python
def run_walk_forward(listings, track, params, model_name):

    safe = data.hide_answers(listings, track)

    if track == "off_market":
        X = features.off_market_features(safe)
    else:
        X = features.list_aware_features(safe)

    results = []
    predictions = []

    for month in validation_months:

        train_mask = (
            listings["date_sold"] < month.start_time
        )

        valid_mask = (
            listings["date_sold"].dt.to_period("M") == month
        )

        X_train = X.loc[train_mask]
        X_valid = X.loc[valid_mask]

        y_train = np.log(
            listings.loc[train_mask, "sold_price"]
        )

        y_valid = listings.loc[
            valid_mask, "sold_price"
        ].to_numpy()

        model = HistGradientBoostingRegressor(**params)

        with threadpool_limits(limits=4):
            model.fit(X_train, y_train)

            predicted = np.round(
                np.exp(model.predict(X_valid))
            )

        ape = np.abs(predicted - y_valid) / y_valid * 100

        results.append({
            "track": track,
            "model": model_name,
            "month": str(month),
            "n": len(y_valid),
            "mdape": np.median(ape),
            "mape": np.mean(ape)
        })

        predictions.append(pd.DataFrame({
            "track": track,
            "model": model_name,
            "month": str(month),
            "ape": ape
        }))

        print(
            f"{track} | {model_name} | {month} | "
            f"MdAPE: {np.median(ape):.3f}% | "
            f"MAPE: {np.mean(ape):.3f}%"
        )

    monthly = pd.DataFrame(results)
    pred_df = pd.concat(predictions, ignore_index=True)

    overall = {
        "track": track,
        "model": model_name,
        "n": len(pred_df),
        "mdape": pred_df["ape"].median(),
        "mape": pred_df["ape"].mean()
    }

    return monthly, overall

print("Walk-forward function ready!")
```

```python
all_monthly = []
all_overall = []

for track, configs in model_configs.items():

    for model_name, params in configs.items():

        print("\n" + "=" * 55)
        print("Running:", track, model_name)
        print("=" * 55)

        monthly, overall = run_walk_forward(
            listings=listings,
            track=track,
            params=params,
            model_name=model_name
        )

        all_monthly.append(monthly)
        all_overall.append(overall)

monthly_results = pd.concat(
    all_monthly,
    ignore_index=True
)

comparison_results = pd.DataFrame(all_overall)

print("\nFINAL COMPARISON")
display(comparison_results)
```
